```python
fleni600_nifti_files_dir = "~/Downloads/Fleni/fleni600/converted_niftis/"
fleni600_nifti_files = os.listdir(
    os.path.expanduser(fleni600_nifti_files_dir)
)
fleni600_nifti_files_ids = [
    os.path.splitext(os.path.splitext(os.path.basename(f))[0])[0].split("_")[1] for f in
    fleni600_nifti_files
]

fleni600_nifti_files_ids = set(fleni600_nifti_files_ids)
fleni600_ids = set(fleni600["pet_id"].astype(str).values)
# len(fleni600_nifti_files_ids), len(fleni600_ids), len(fleni600_nifti_files_ids.intersection(fleni600_ids))
```

```python
# list contents of zips

import zipfile

def list_zip_contents(zip_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        return zip_ref.namelist()

fleni600_zip_path = "~/Downloads/Fleni/fleni600/"

fleni600_zips = glob.glob(os.path.expanduser(fleni600_zip_path + "*.zip"))

zips_n_contents = {
    zip_file.split("/")[-1].split(".")[0]: (
        list({file.split("/")[0] for file in list_zip_contents(zip_file)})[0]
    ).split("Anonymized")[1]
    for zip_file in fleni600_zips
}

# bi directional mapping
file2id = {
    file: id_ for id_, file in zips_n_contents.items()
}
```

```python
len(set(file2id.values()).intersection(fleni600_ids)), len(fleni600_ids), len(set(file2id.values()))
```

```python
fleni600["zip_id"] = fleni600["pet_id"].map(zips_n_contents)
fleni600
```

```python
# "subject_1235_pet.nii.gz" example file
# dataset loading os.path.join(self.data_dir, f"{subject_id}.nii.gz"),

fleni600["Subject"] = fleni600["zip_id"].apply(
    lambda x: f"subject_{x}_pet"
)

fleni600
```
