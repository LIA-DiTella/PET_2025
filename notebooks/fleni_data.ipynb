```python
fleni100_nifti_files_dir = "~/Downloads/Fleni/fleni100/entelai-fleni-pet-base-diferencial/panania/Neuro/fleni_pet_en_base_dif/fleni100/"

fleni100_nifti_files = os.listdir(
    os.path.expanduser(fleni100_nifti_files_dir)
)

fleni100_nifti_files_ids = [
    os.path.splitext(os.path.splitext(os.path.basename(f))[0])[0] for f in fleni100_nifti_files
]

fleni100_nifti_files_ids = set(fleni100_nifti_files_ids)

fleni100_ids = set(fleni100["pet_id"].values)

len(fleni100_nifti_files_ids), len(fleni100_ids), len(fleni100_nifti_files_ids.intersection(fleni100_ids))
```
